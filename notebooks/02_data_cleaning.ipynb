# 02 - Data Cleaning

I drop the 3 constant columns and the ID column, then build the 1/0 target flag.

```python
import pandas as pd

employees = pd.read_csv('data/raw/WA_Fn-UseC_-HR-Employee-Attrition.csv')
constant_columns = [c for c in employees.columns if employees[c].nunique() == 1]
columns_to_drop = constant_columns + ['EmployeeNumber']
hr = employees.drop(columns=columns_to_drop).copy()
print(employees.shape, '->', hr.shape)
```

## Target flag

The mean of a 0/1 column equals the attrition rate.

```python
hr['AttritionFlag'] = (hr['Attrition'] == 'Yes').astype(int)
print('Attrition rate: %.1f%%' % (hr['AttritionFlag'].mean() * 100))
```

## Feature engineering - readable bands

```python
hr['AgeBand'] = pd.cut(hr['Age'], bins=[17,25,35,45,55,65], labels=['18-25','26-35','36-45','46-55','56+'])
hr['TenureBand'] = pd.cut(hr['YearsAtCompany'], bins=[-1,1,3,5,10,40], labels=['0-1 yrs','2-3 yrs','4-5 yrs','6-10 yrs','11+ yrs'])
hr['DistanceBand'] = pd.cut(hr['DistanceFromHome'], bins=[-1,2,9,19,29], labels=['0-2','3-9','10-19','20+'])
hr['IncomeBand'] = pd.cut(hr['MonthlyIncome'], bins=[0,3000,6000,10000,20000], labels=['<3k','3k-6k','6k-10k','10k+'])
hr.to_csv('data/processed/hr_cleaned.csv', index=False)
hr.shape
```
