# 01 - Data Understanding

What is inside the raw file before I change anything?

```python
import pandas as pd

RAW_FILE = 'data/raw/WA_Fn-UseC_-HR-Employee-Attrition.csv'
employees = pd.read_csv(RAW_FILE)
print('Rows:', employees.shape[0], 'Columns:', employees.shape[1])
```

## Missing values and duplicates

```python
print('Missing values:', employees.isnull().sum().sum())
print('Duplicate rows:', employees.duplicated().sum())
```

## Constant columns carry no information

I prove a column is constant with `nunique()` instead of assuming it.

```python
constant_columns = [c for c in employees.columns if employees[c].nunique() == 1]
constant_columns
```

## The target variable: how many employees left?

```python
employees['Attrition'].value_counts()
```
