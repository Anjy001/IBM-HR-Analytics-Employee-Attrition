# 04 - Statistical Analysis

I use chi-square for categories and Welch t-tests for numbers, with effect sizes alongside.

```python
import numpy as np
import pandas as pd
from scipy import stats

hr = pd.read_csv('data/processed/hr_cleaned.csv')

def chi_square_test(data, column):
    crosstab = pd.crosstab(data[column], data['AttritionFlag'])
    chi2, p, dof, expected = stats.chi2_contingency(crosstab)
    n = crosstab.values.sum()
    k = min(crosstab.shape) - 1
    return {'chi2': round(chi2, 2), 'p_value': round(p, 6), 'cramers_v': round(np.sqrt(chi2/(n*k)), 3)}

def ttest_numeric(data, column):
    leavers = data.loc[data['AttritionFlag'] == 1, column]
    stayers = data.loc[data['AttritionFlag'] == 0, column]
    t, p = stats.ttest_ind(leavers, stayers, equal_var=False)
    return {'leaver_mean': round(leavers.mean(), 2), 'stayer_mean': round(stayers.mean(), 2), 't': round(t, 2), 'p_value': round(p, 6)}
```

## Chi-square: is the variable independent of attrition?

```python
for column in ['OverTime', 'JobRole', 'Gender', 'PerformanceRating']:
    print(column, chi_square_test(hr, column))
```

## T-tests: do leavers and stayers differ?

```python
for column in ['MonthlyIncome', 'Age', 'YearsAtCompany', 'DistanceFromHome']:
    print(column, ttest_numeric(hr, column))
```
