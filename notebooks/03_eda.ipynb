# 03 - Exploratory Data Analysis

I keep the group size next to every rate.

```python
import pandas as pd

hr = pd.read_csv('data/processed/hr_cleaned.csv')

def attrition_table(data, column):
    table = data.groupby(column, observed=True).agg(
        Employees=('AttritionFlag', 'size'),
        Leavers=('AttritionFlag', 'sum'),
        AttritionRate=('AttritionFlag', 'mean'))
    table['AttritionRate'] = (table['AttritionRate'] * 100).round(1)
    return table.sort_values('AttritionRate', ascending=False)

print('Attrition rate: %.1f%%' % (hr['AttritionFlag'].mean() * 100))
```

## By job role

```python
attrition_table(hr, 'JobRole')
```

## By overtime

```python
attrition_table(hr, 'OverTime')
```

## The cut from the dataset page: distance from home by role x attrition

```python
hr.pivot_table(index='JobRole', columns='Attrition', values='DistanceFromHome', aggfunc='mean').round(1)
```

## The second cut: income by education x attrition

```python
hr.pivot_table(index='Education', columns='Attrition', values='MonthlyIncome', aggfunc='mean').round(0)
```
